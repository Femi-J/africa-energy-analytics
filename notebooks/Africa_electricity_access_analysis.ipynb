## My Analysis — Key Findings

Looking at electricity access data across 16 African countries in 2023, the most surprising finding was Rwanda's remarkable 18.7% improvement in just 3 years, suggesting that deliberate government policy can  dramatically accelerate energy access even in low-income countries.

South Africa's decline of 2.3% was equally striking. As the most industrialised economy in the sample, its regressions show how infrastructure neglect can reverse hard-earned progress.

Nigeria, despite being Africa's largest economy, sits at only 61.2% access. A gap that represents both a crisis and an enormous opportunity for targeted investment in renewable energy solutions.

Across the 16 countries, the average electricity access rate stood at 67.2% in 2023, meaning that roughly one in every three people still lacked access to electricity, showing the magnitude of the challenge that remains.

```python
import matplotlib.pyplot as plt

# Sort by 2023 access for cleaner visualisation
latest_sorted = africa_df[['Country Name', '2023']].dropna().sort_values('2023', ascending=True)

# Calculate average
avg = africa_df['2023'].mean()

# Create the chart
fig, ax = plt.subplots(figsize=(10, 8))

# Plot bars
bars = ax.barh(
    latest_sorted['Country Name'],
    latest_sorted['2023'],
    color='#00c27c',
    edgecolor='none',
    height=0.6
)

# Highlight bars below average in a different colour
for bar, value in zip(bars, latest_sorted['2023']):
    if value < avg:
        bar.set_color('#ff6b6b')

# Add value labels on each bar
for bar, value in zip(bars, latest_sorted['2023']):
    ax.text(value + 0.5, bar.get_y() + bar.get_height()/2,
            f'{value:.1f}%', va='center', fontsize=9)

# Add average line
ax.axvline(x=avg, color='orange', linestyle='--', linewidth=1.5,
           label=f'Average: {avg:.1f}%')

# Styling
ax.set_title('Electricity Access by Country (2023)\nGreen = Above Average | Red = Below Average',
             fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('% of Population with Electricity Access', fontsize=11)
ax.set_xlim(0, 115)
ax.legend(fontsize=10)
ax.spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.savefig('../reports/electricity_access_2023.png', dpi=150)
plt.show()

print("Chart saved to /reports folder!")
```

```python
# Filtering with multiple conditions
print("=== Countries with access ABOVE 60% AND improved since 2020 ===")
africa_df_copy = africa_df.copy()
africa_df_copy['improvement'] = africa_df_copy['2023'] - africa_df_copy['2020']

filtered = africa_df_copy[
    (africa_df_copy['2023'] > 60) & 
    (africa_df_copy['improvement'] > 0)
]
print(filtered[['Country Name', '2020', '2023', 'improvement']])
```

```python
# Add a region column to group countries by
region_map = {
    'Nigeria': 'West Africa',
    'Ghana': 'West Africa',
    'Senegal': 'West Africa',
    "Cote d'Ivoire": 'West Africa',
    'Cameroon': 'Central Africa',
    'Kenya': 'East Africa',
    'Ethiopia': 'East Africa',
    'Tanzania': 'East Africa',
    'Uganda': 'East Africa',
    'Rwanda': 'East Africa',
    'South Africa': 'Southern Africa',
    'Zambia': 'Southern Africa',
    'Namibia': 'Southern Africa',
    'Malawi': 'Southern Africa',
    'Egypt, Arab Rep.': 'North Africa',
    'Morocco': 'North Africa'
}

africa_df_copy['Region'] = africa_df_copy['Country Name'].map(region_map)
print(africa_df_copy[['Country Name', 'Region', '2023']].to_string(index=False))
```

```python
# Group by region and calculate average electricity access
regional_avg = africa_df_copy.groupby('Region')['2023'].agg(['mean', 'min', 'max']).round(1)
regional_avg.columns = ['Average %', 'Lowest %', 'Highest %']
regional_avg = regional_avg.sort_values('Average %', ascending=False)
print("=== Electricity Access by Region (2023) ===")
print(regional_avg.to_string())
```

```python
# Check for missing values in key year columns
print("=== Missing Values in Year Columns ===")
year_cols = ['2020', '2021', '2022', '2023']
print(africa_df_copy[year_cols].isnull().sum())

# Fill missing values with the row average
africa_df_copy[year_cols] = africa_df_copy[year_cols].apply(
    lambda row: row.fillna(row.mean()), axis=1
)
print("\nMissing values after filling:")
print(africa_df_copy[year_cols].isnull().sum())
```

## Analysis 2 — 3 Key Questions

```python
# Q1: Which region has the biggest electricity access gap to close by 2030?
print("=== Regional Gap to Universal Access (100%) ===")
regional_avg['Gap to 100%'] = 100 - regional_avg['Average %']
print(regional_avg[['Average %', 'Gap to 100%']].sort_values('Gap to 100%', ascending=False))
```

```python
# Q2: Which country has the most consistent steady growth?
print("=== Year on Year Growth Consistency ===")
africa_df_copy['YoY_2021'] = africa_df_copy['2021'] - africa_df_copy['2020']
africa_df_copy['YoY_2022'] = africa_df_copy['2022'] - africa_df_copy['2021']
africa_df_copy['YoY_2023'] = africa_df_copy['2023'] - africa_df_copy['2022']
africa_df_copy['Avg_YoY'] = africa_df_copy[['YoY_2021','YoY_2022','YoY_2023']].mean(axis=1).round(2)
print(africa_df_copy[['Country Name', 'YoY_2021', 'YoY_2022', 'YoY_2023', 'Avg_YoY']]
      .sort_values('Avg_YoY', ascending=False).to_string(index=False))
```

```python
# Q3: Which region is improving fastest on average?
print("=== Regional Improvement 2020 to 2023 ===")
africa_df_copy['improvement'] = africa_df_copy['2023'] - africa_df_copy['2020']
regional_improvement = africa_df_copy.groupby('Region')['improvement'].mean().round(2)
print(regional_improvement.sort_values(ascending=False))
```

## Key findings from analsis of the 3 Key Questions

### Q1 - Regional Gap to Universal Access

Southern Africa shows the largest access gap at 47.2%, making it the most structurally underserved region on paper. However, this figure is heavily influenced by outliers such as Malawi, which has significantly lower access levels compared to regional peers.

More importantly, East Africa’s 40.9% gap represents the more systemic challenge. Unlike Southern Africa, East Africa combines a large population base with widespread access deficits, meaning the absolute number of people affected is substantially higher.

From an investment perspective, East and Southern Africa represent the highest-impact electrification frontiers, with East Africa carrying the greater scale urgency and Southern Africa reflecting deeper intra-regional inequality.

### Q2 - Year-on-Year Growth Consistency

Rwanda is the standout performer with an average annual growth rate of 6.23%, reinforcing its position as a high-velocity electrification leader. However, the 2023 spike of 13.3% signals a concentrated acceleration event, likely driven by a targeted policy push or infrastructure expansion phase rather than uniform structural growth.

In contrast, Tanzania demonstrates the strongest consistency profile (2.8%, 3.1%, 2.5%). While its growth rate is lower than Rwanda’s peak, its stability indicates a predictable and sustained electrification pathway, which is often more replicable and scalable for policy design.

At the other end of the spectrum, South Africa records a negative average growth (-0.77%), making it the only country in the dataset showing net regression. This is a critical signal of stagnation or system stress within a historically advanced electricity market.

### Q3 - Fastest Improving Region

East Africa leads regional improvement with an aggregate gain of 9.1% between 2020 and 2023, driven primarily by strong performance from Rwanda and Uganda.

Central Africa follows closely, indicating broad-based but less intense progress, while West Africa shows moderate gains.

Notably, North Africa registers 0% improvement, not due to underperformance, but because it has already reached full electrification (100%), placing it at a natural saturation ceiling rather than a growth failure point.

```python
import pandas as pd

wb_df = pd.read_csv('../data/API_EG.ELC.ACCS.ZS_DS2_en_csv_v2_127016.csv',
                    skiprows=4)
print("World Bank data loaded successfully")
print(f"Shape: {wb_df.shape}")
```

```python
# Map solar study countries to World Bank country names
country_name_map = {
    'Nigeria':       'Nigeria',
    'Ghana':         'Ghana',
    'Senegal':       'Senegal',
    "Cote d'Ivoire": "Cote d'Ivoire",
    'Mali':          'Mali',
    'Burkina Faso':  'Burkina Faso',
    'Niger':         'Niger',
    'Guinea':        'Guinea',
    'Togo':          'Togo',
    'Benin':         'Benin',
    'Sierra Leone':  'Sierra Leone',
    'Liberia':       'Liberia',
    'Mauritania':    'Mauritania',
    'Gambia':        'Gambia, The',
    'Guinea-Bissau': 'Guinea-Bissau',
    'Cape Verde':    'Cabo Verde',
}

# Filter World Bank data for these 16 countries
wa_16 = wb_df[wb_df['Country Name'].isin(country_name_map.values())]

# Clean display names
display_names = {v: k for k, v in country_name_map.items()}
wa_16 = wa_16.copy()
wa_16['Display Name'] = wa_16['Country Name'].map(display_names)

# Sort by 2023 access
wa_16 = wa_16.sort_values('2023', ascending=True)
avg_wa16 = wa_16['2023'].mean()

print(f"Countries found: {len(wa_16)}")
print(f"Average access: {avg_wa16:.1f}%")
print(f"Lacking access: {100 - avg_wa16:.1f}%")
print(wa_16[['Display Name', '2023']].to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(10, 8))

bars = ax.barh(
    wa_16['Display Name'],
    wa_16['2023'],
    color=['#ff6b6b' if v < avg_wa16 else '#00c27c'
           for v in wa_16['2023']],
    edgecolor='none',
    height=0.6
)

# Value labels
for bar, val in zip(bars, wa_16['2023']):
    ax.text(val + 0.5, bar.get_y() + bar.get_height()/2,
            f'{val:.1f}%', va='center', fontsize=9)

# Average line
ax.axvline(x=avg_wa16, color='#333333', linestyle='--',
           linewidth=1.5, label=f'Regional average: {avg_wa16:.1f}%')

# Styling
ax.set_title('Figure 1 — Electricity Access by Country\n16 West African Nations (2023)',
             fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('% Population with Electricity Access', fontsize=11)
ax.set_xlim(0, 115)
ax.legend(fontsize=10)
ax.spines[['top', 'right']].set_visible(False)

# Add annotation
ax.text(56, -1.2,
        'Green = above regional average  |  Red = below regional average',
        fontsize=8, color='#666666', ha='center')

plt.tight_layout()
plt.savefig('../reports/fig1_wa_electricity_access_2023.png', dpi=300)
plt.show()
print("Figure 1 saved!")
```
